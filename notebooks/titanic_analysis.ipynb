Goal:
1) Use **k-means clustering** to describe passenger groupings.
2) Use **logistic regression** to predict who might survive.

Modeling fields:
- Numeric: `pclass, age, sibsp, parch, fare`
- Categorical: `sex, embarked`

Excluded (by design): high-cardinality text (`name`, `ticket`, `cabin`, `home.dest`) and leakage (`boat`, `body`).

```
DATA_PATH = 'CS379T-Week-1-IP.xls'  # place the dataset in the repository root
OUT_DIR = 'outputs'

import sys
from pathlib import Path
from IPython.display import display, Image

sys.path.insert(0, str(Path('src').resolve()))
Path(OUT_DIR).mkdir(exist_ok=True)
```

## 1) Load + quick inspection

```
from titanic_clustering import load_titanic

df = load_titanic(DATA_PATH)
display(df.head())
display((df.isna().mean().sort_values(ascending=False) * 100).round(2).rename('% missing').head(12))
display(df[['survived','pclass','sex','age','sibsp','parch','fare','embarked']].describe(include='all'))
```

## 2) Unsupervised: k-means clustering
Rationale: k-means is a simple baseline to discover group structure; silhouette score helps select k.
Capture screenshots of the table and plots below for your report.

```
from titanic_clustering import run as run_clustering

best_k, sil_by_k, cluster_profile = run_clustering(DATA_PATH, OUT_DIR)
print('Silhouette scores:', sil_by_k)
print('Selected k:', best_k)
display(cluster_profile)

Image(filename=str(Path(OUT_DIR) / 'silhouette_scores.png'))
```

```
Image(filename=str(Path(OUT_DIR) / 'clusters_pca.png'))
```

## 3) Supervised: logistic regression survival prediction
Rationale: logistic regression is a straightforward classifier and its coefficients are interpretable.
Capture screenshots of the report and plots below for your report.

```
from titanic_classification import run as run_classification

results = run_classification(DATA_PATH, OUT_DIR)
print(results['classification_report'])
print('ROC-AUC:', results['roc_auc'])
print('\nTop coefficients (feature -> weight):')
for name, w in results['top_coefficients']:
    print(f'  {name}: {w:.3f}')

Image(filename=str(Path(OUT_DIR) / 'confusion_matrix.png'))
```

```
Image(filename=str(Path(OUT_DIR) / 'roc_curve.png'))
```
